```python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
    else:
        print("Result:", result)
    finally:
        print("Execution completed")


divide(10, 2)
```

```python
def convert_integer(value):
    try:
        num = int(value)
    except ValueError:
        print("Invalid input")
    else:
        print("Converted value:", num)
    finally:
        print("Conversion attempt finished")


convert_integer("25")
```

```python
def square_number(value):
    try:
        num = float(value)
        square = num ** 2
    except ValueError:
        print("Invalid input")
    else:
        print("Square:", square)
        return square
    finally:
        print("Process done")


square_number("5.5")
```

```python
def floor_division(a, b):
    try:
        quotient = a // b
    except ZeroDivisionError:
        print("Cannot divide by zero")
    else:
        print("Quotient:", quotient)
    finally:
        print("Function executed")


floor_division(10, 3)
```

```python
def multiply(a, b):
    try:
        num1 = int(a)
        num2 = int(b)
        product = num1 * num2
    except ValueError:
        print("Invalid input")
    else:
        print("Product:", product)
    finally:
        print("Multiplication tried")


multiply("10", "5")
```

```python
def percentage(marks, total_marks):
    try:
        result = (marks / total_marks) * 100
    except ZeroDivisionError:
        print("Total marks cannot be zero")
    else:
        print("Percentage:", result, "%")
    finally:
        print("Calculation finished")


percentage(450, 500)
```

```python
def half_value(value):
    try:
        num = float(value)
        half = num / 2
    except ValueError:
        print("Invalid input")
    else:
        print("Half:", half)
    finally:
        print("Conversion completed")


half_value("20.5")
```

```python
def modulus(a, b):
    try:
        result = a % b
    except Exception as e:
        print("Error:", e)
    else:
        print("Remainder:", result)
    finally:
        print("Modulus attempt done.")


modulus(10, 3)
```

```python
def divide_three(a, b, c):
    try:
        result1 = a / b
        result2 = result1 / c
    except Exception as e:
        print("Error:", e)
    else:
        print("Final result:", result2)
    finally:
        print("Division process ended.")


divide_three(100, 5, 2)
```

```python
def double_number(num):
    try:
        num = int(num)
        result = num * 2
    except Exception as e:
        print("Error:", e)
    else:
        print("Double:", result)
    finally:
        print("Closing message: Conversion completed.")


double_number("25")
```

```python
def subtraction(a, b):
    try:
        a = float(a)
        b = float(b)
        result = a - b
    except Exception as e:
        print("Error:", e)
    else:
        print("Difference:", result)
    finally:
        print("Subtraction attempt finished.")


subtraction("50", "20")
```

```python
def average(a, b):
    try:
        result = (a + b) / 2
    except Exception as e:
        print("Error:", e)
    else:
        print("Average:", result)
    finally:
        print("Average calculation done.")


average(10, 20)
```
